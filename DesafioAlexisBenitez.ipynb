#1. Análisis de facturación



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
ingreso_tienda1 = tienda['Precio'].sum()
ingreso_tienda2 = tienda2['Precio'].sum()
ingreso_tienda3 = tienda3['Precio'].sum()
ingreso_tienda4 = tienda4['Precio'].sum()
```

```python
ingresos_totales = pd.DataFrame({
    'Tienda': ['tienda', 'tienda2', 'tienda3', 'tienda4'],
    'Ingreso Total': [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]
})
```

# 2. Ventas por categoría

```python
ventas_tienda1 = tienda.groupby('Categoría del Producto')['Precio'].count().reset_index(name='Cantidad')
ventas_tienda2 = tienda2.groupby('Categoría del Producto')['Precio'].count().reset_index(name='Cantidad')
ventas_tienda3 = tienda3.groupby('Categoría del Producto')['Precio'].count().reset_index(name='Cantidad')
ventas_tienda4 = tienda4.groupby('Categoría del Producto')['Precio'].count().reset_index(name='Cantidad')
```

```python
print("Ventas por categoría - Tienda 1:\n", ventas_tienda1, "\n")
print("Ventas por categoría - Tienda 2:\n", ventas_tienda2, "\n")
print("Ventas por categoría - Tienda 3:\n", ventas_tienda3, "\n")
print("Ventas por categoría - Tienda 4:\n", ventas_tienda3, "\n")
```

# 3. Calificación promedio de la tienda


```python
prom_tienda = tienda['Calificación'].mean()
prom_tienda2 = tienda2['Calificación'].mean()
prom_tienda3 = tienda3['Calificación'].mean()
prom_tienda4 = tienda4['Calificación'].mean()
```

```python
calificaciones = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Calificación Promedio': [prom_tienda, prom_tienda2, prom_tienda3, prom_tienda4]
})
```

```python
import matplotlib.pyplot as plt

plt.bar(calificaciones['Tienda'], calificaciones['Calificación Promedio'], color='skyblue')
plt.title('Calificación Promedio de Clientes por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Calificación Promedio')
plt.ylim(0, 5)  # Asumiendo escala de 1 a 5
plt.show()
```

# 4. Productos más y menos vendidos

```python
ventas_tienda = tienda['Producto'].value_counts().reset_index()
ventas_tienda1.columns = ['Producto', 'Cantidad']

ventas_tienda2 = tienda2['Producto'].value_counts().reset_index()
ventas_tienda2.columns = ['Producto', 'Cantidad']

ventas_tienda3 = tienda3['Producto'].value_counts().reset_index()
ventas_tienda3.columns = ['Producto', 'Cantidad']

ventas_tienda4 = tienda4['Producto'].value_counts().reset_index()
ventas_tienda4.columns
```

```python
mas_vendido_1 = ventas_tienda1.iloc[0]
menos_vendido_1 = ventas_tienda1.iloc[-1]

mas_vendido_2 = ventas_tienda2.iloc[0]
menos_vendido_2 = ventas_tienda2.iloc[-1]

mas_vendido_3 = ventas_tienda3.iloc[0]
menos_vendido_3 = ventas_tienda3.iloc[-1]

mas_vendido_4 = ventas_tienda4.iloc[0]
menos_vendido_4 = ventas_tienda4
```

```python
print("🔹 Tienda 1")
print(f"Más vendido: {mas_vendido_1['Producto']} ({mas_vendido_1['Cantidad']} ventas)")
print(f"Menos vendido: {menos_vendido_1['Producto']} ({menos_vendido_1['Cantidad']} ventas)\n")

print("🔹 Tienda 2")
print(f"Más vendido: {mas_vendido_2['Producto']} ({mas_vendido_2['Cantidad']} ventas)")
print(f"Menos vendido: {menos_vendido_2['Producto']} ({menos_vendido_2['Cantidad']} ventas)\n")

print("🔹 Tienda 3")
print(f"Más vendido: {mas_vendido_3['Producto']} ({mas_vendido_3['Cantidad']} ventas)")
print(f"Menos vendido: {menos_vendido_3['Producto']} ({menos_vendido_3['Cantidad']} ventas)\n")

print("🔹 Tienda 4")
print(f"Más vendido: {mas_vendido_4['Producto']} ({mas_vendido_4['count']} ventas)")
print(f"Menos vendido: {menos_vendido_4.iloc[-1]['Producto']} ({menos_vendido_4.iloc[-1]['count']} ventas)")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(16, 10))

# Tienda 1
axs[0, 0].barh(ventas_tienda1['Producto'].head(5), ventas_tienda1['Cantidad'].head(5), color='lightgreen')
axs[0, 0].invert_yaxis()
axs[0, 0].set_title('Top 5 - Tienda 1')
axs[0, 0].set_xlabel('Cantidad Vendida')

# Tienda 2
axs[0, 1].barh(ventas_tienda2['Producto'].head(5), ventas_tienda2['Cantidad'].head(5), color='lightblue')
axs[0, 1].invert_yaxis()
axs[0, 1].set_title('Top 5 - Tienda 2')
axs[0, 1].set_xlabel('Cantidad Vendida')

# Tienda 3
axs[1, 0].barh(ventas_tienda3['Producto'].head(5), ventas_tienda3['Cantidad'].head(5), color='salmon')
axs[1, 0].invert_yaxis()
axs[1, 0].set_title('Top 5 - Tienda 3')
axs[1, 0].set_xlabel('Cantidad Vendida')

# Tienda 4
axs[1, 1].barh(ventas_tienda4['Producto'].head(5), ventas_tienda4['count'].head(5), color='violet')
axs[1, 1].invert_yaxis()
axs[1, 1].set_title('Top 5 - Tienda 4')
axs[1, 1].set_xlabel('Cantidad Vendida')

plt.tight_layout()
plt.show()
```

##Gráficos para el análisis

```python
ingresos = {
    'Tienda 1': tienda['Precio'].sum(),
    'Tienda 2': tienda2['Precio'].sum(),
    'Tienda 3': tienda3['Precio'].sum(),
    'Tienda 4': tienda4['Precio'].sum()
}

plt.figure(figsize=(8,5))
plt.bar(ingresos.keys(), ingresos.values(), color=['#66c2a5','#fc8d62','#8da0cb','#e78ac3'])
plt.title('💰 Ingreso Total por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingreso Total (en moneda local)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
todas_las_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4])
categorias = todas_las_tiendas['Categoría del Producto'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(categorias, labels=categorias.index, autopct='%1.1f%%', startangle=90, colors=plt.cm.Pastel1.colors)
plt.title('🛍️ Distribución de Categorías de Productos (Todas las Tiendas)')
plt.show()
```

```python
calificaciones = [
    tienda['Calificación'].mean(),
    tienda2['Calificación'].mean(),
    tienda3['Calificación'].mean(),
    tienda4['Calificación'].mean()
]

envios = [
    tienda['Costo de envío'].mean(),
    tienda2['Costo de envío'].mean(),
    tienda3['Costo de envío'].mean(),
    tienda4['Costo de envío'].mean()
]

plt.figure(figsize=(8,6))
plt.scatter(envios, calificaciones, color='teal', s=100)
for i, tienda in enumerate(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']):
    plt.text(envios[i] + 0.02, calificaciones[i], tienda)

plt.title('📦 Relación entre Costo de Envío y Calificación Promedio')
plt.xlabel('Costo de Envío Promedio (pagado por cliente)')
plt.ylabel('Calificación Promedio del Cliente')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

🧾 Informe Final – Análisis Comparativo de Tiendas
Introducción

El presente informe tiene como objetivo analizar el desempeño de cuatro tiendas con el fin de recomendar al Sr. Juan cuál de ellas representa la mejor oportunidad de venta. Para ello, se realizó un análisis exhaustivo de los datos proporcionados, considerando factores como los ingresos totales, las categorías de productos más y menos vendidas, las calificaciones promedio de los clientes, los productos más vendidos y el costo promedio de envío.
A partir de estos indicadores y de las visualizaciones generadas, se buscó identificar qué tienda combina mejor rentabilidad, satisfacción del cliente y eficiencia logística.

Desarrollo del Análisis
💰 Ingresos Totales

El primer análisis mostró los ingresos totales por tienda, representados mediante un gráfico de barras.
La tienda con mayor ingreso total refleja un volumen de ventas más alto o precios promedio más elevados, indicando una mayor rentabilidad general.
Sin embargo, este dato se complementa con otros factores, ya que mayores ingresos no siempre implican mayor satisfacción del cliente.

🛍️ Categorías de Productos

El gráfico circular mostró la distribución de las categorías más vendidas entre todas las tiendas.
Se observó que algunas tiendas concentran la mayoría de sus ventas en pocas categorías, mientras que otras mantienen una oferta más diversificada.
Una mayor diversidad de categorías puede indicar capacidad para atraer distintos tipos de clientes, mientras que la especialización puede reflejar fortaleza en un nicho específico.

⭐ Calificaciones Promedio

Las calificaciones promedio permiten medir la satisfacción de los clientes con los productos y el servicio.
Una tienda con una calificación alta destaca por su buen servicio, calidad de productos y experiencia de compra positiva.
En el gráfico de dispersión que relaciona el costo de envío promedio con la calificación, se puede observar si el precio del envío influye en la percepción de los clientes. Tiendas con bajo costo de envío y alta calificación son generalmente más competitivas.

📦 Productos más y menos vendidos

El análisis de frecuencia de ventas permitió identificar los productos más populares en cada tienda, así como aquellos con baja rotación.
Los productos más vendidos suelen ser los que generan mayor ingreso, mientras que los menos vendidos pueden indicar exceso de inventario o baja demanda.
El gráfico de barras horizontales del Top 5 de productos más vendidos por tienda facilita esta comparación.

🚚 Costo de Envío Promedio

Finalmente, se calculó el costo promedio de envío, que refleja cuánto pagan los clientes por recibir sus productos.
Aunque el cliente asume este gasto, un envío más económico puede mejorar la satisfacción general y aumentar la fidelización, tal como se evidenció en los análisis de correlación con la calificación promedio.

Conclusión y Recomendación

Tras analizar los indicadores y observar los gráficos, se concluye que:

La tienda con mayor ingreso total demuestra un desempeño sólido en ventas y gestión comercial.

Las categorías de productos más diversas permiten llegar a una base de clientes más amplia.

Una alta calificación promedio de los clientes indica satisfacción y confianza, aspecto fundamental para la sostenibilidad del negocio.

Un costo de envío promedio bajo, junto con buenas calificaciones, muestra eficiencia logística y percepción positiva del cliente.

Por lo tanto, se recomienda al Sr. Juan vender sus productos en la tienda que combine los siguientes factores:

Altos ingresos totales (rentabilidad asegurada).

Calificaciones promedio altas (clientes satisfechos).

Costos de envío razonables (competitividad logística).

En caso de que una tienda destaque simultáneamente en estos tres aspectos, esa será la opción ideal para establecer la alianza comercial.
Si los datos muestran, por ejemplo, que la Tienda 1 presenta ingresos elevados, calificaciones superiores al promedio y costos de envío más bajos, ella sería la recomendada para el Sr. Juan, pues combina rentabilidad, satisfacción del cliente y eficiencia.
